BRANDS WITH MOST TWEETS

```python
candidate_brands = (
    support_tweets["author_id"]
    .value_counts()
    .head(10)
)

candidate_brands
```

```python
brands = candidate_brands.index.tolist()

brand_stats = []

for brand in brands:
    brand_support = df[
        (df["author_id"] == brand) &
        (df["inbound"] == False)
    ]

    brand_stats.append({
        "brand": brand,
        "support_tweets": len(brand_support)
    })

brand_stats = pd.DataFrame(brand_stats)

brand_stats
```
